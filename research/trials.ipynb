```python
from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def load_pdf_data(data):
    loader = DirectoryLoader(
        data,
        glob ="*.pdf",
        loader_cls=PyPDFLoader
    )
    documents = loader.load()
    return documents
```

```python
extract_data = load_pdf_data(data="../data/")
```

```python
extract_data
```

```python
from typing import List
from langchain_core.documents import Document
```

```python
def filter_with_minimal_docs(docs: List[Document]) ->List[Document]:
    minimal_docs: List[Document] =[]
    for doc in docs:
        src = doc.metadata.get("source")
        minimal_docs.append(
            Document(
                page_content=doc.page_content,
                metadata={"source":src}

            )
        )
    return minimal_docs
```

```python
minimal_docs = filter_with_minimal_docs(extract_data)
```

```python
minimal_docs
```

```python
def text_split(minimal_docs):
    text_split = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
    text_chunk = text_split.split_documents(minimal_docs)
    return text_chunk
```

```python
text_chunk = text_split(minimal_docs)
print(f"Total chunk: {len(text_chunk)}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

def huggingface_embedding():
    embedding = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
    return embedding
```

```python
embedding = huggingface_embedding()

result = embedding.embed_query("Hello! I am Sifat")
print(f"Total result: {len(result)}")
```

```python
from pinecone import Pinecone, ServerlessSpec
from langchain_pinecone import PineconeVectorStore

pinecone_api_key = PINECONE_API_KEY
index_name = "agriculture-advisor"
pc = Pinecone(api_key = pinecone_api_key )

if not pc.has_index(index_name):
    pc.create_index(
        name = index_name,
        dimension = 384,
        metric = "cosine",
        spec= ServerlessSpec(cloud="aws",region="us-east-1")
    )
index = pc.Index(index_name)
```

```python
vectorestore = PineconeVectorStore.from_existing_index(
    index_name = index_name,
    embedding = embedding
)

stats = vectorestore.index.describe_index_stats()

vector_count = (
    stats.get("namespaces",{})
        .get("",{})
        .get("vector_count",0)
)

if vector_count == 0:
    vectorestore = PineconeVectorStore.from_documents(
       documents= text_chunk,
       index_name = index_name,
       embedding = embedding
    )
```

```python
retriever = vectorestore.as_retriever(
    search_type = "similarity",
    search_kwargs = {"k":3}
)
```

```python
from langchain_groq import ChatGroq

chat_model = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature = 0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.output_parsers import StrOutputParser
```

```python
SYSTEM_PROMPT = """
You are an agricultural information assistant.

Your role is to provide accurate, educational agricultural information
based ONLY on the provided context.

CONVERSATION HISTORY:
{history}

AGRICULTURAL CONTEXT:
{context}

STRICT RULES:
- ফসলের রোগ নির্ণয় (diagnosis) নিশ্চিতভাবে ঘোষণা করবে না।
- বিপজ্জনক রাসায়নিকের নির্দিষ্ট মাত্রা বা ডোজ নির্দেশ করবে না।
- কৃষি বিশেষজ্ঞের পরামর্শ ছাড়া চূড়ান্ত সিদ্ধান্ত দিতে উৎসাহিত করবে না।
- যদি তথ্য context-এ না থাকে, বলবে: "প্রদত্ত তথ্যের ভিত্তিতে আমি জানি না।"
- প্রয়োজন হলে স্থানীয় কৃষি অফিস/কৃষি বিশেষজ্ঞের সাথে পরামর্শ করার পরামর্শ দেবে।

STYLE RULES:
- উত্তর সংক্ষিপ্ত, তথ্যভিত্তিক ও স্পষ্ট হবে।
- সর্বোচ্চ ৩টি বাক্য।
- অনুমানভিত্তিক বা অতিরঞ্জিত কথা বলা যাবে না।
- প্রশ্ন ও উত্তরের ভাষা অবশ্যই বাংলা হবে।

Question: {question}
"""
```

```python
prompt = ChatPromptTemplate.from_template(SYSTEM_PROMPT)
```

```python
store = {}

def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

def build_context(question: str):
    docs = retriever.invoke(question)
    return "\n".join(doc.page_content for doc in docs)

def format_history(message):
    if not message:
        return "No previous conversation."
    return "\n".join(f"{m.type.capitalize()}: {m.content}" for m in message)
```

```python
from langchain_core.runnables import RunnableLambda

base_chain = (
    {
        "question": RunnableLambda(lambda x:x["question"]),
        "context": RunnableLambda(lambda x: build_context(x["question"])),
        "history": RunnableLambda(lambda x: format_history(x["history"]))
    }
    | prompt
    | chat_model
    | StrOutputParser()
)
```

```python
agricultural_chain = RunnableWithMessageHistory(
    base_chain,
    get_history,
    input_messages_key="question",
    history_messages_key="history"
)
```

```python
response = agricultural_chain.invoke(
    {"question": "ধান গাছে ব্লাস্ট রোগ কীভাবে সনাক্ত করা যায় এবং প্রতিরোধের উপায় কী?"},
    config={"configurable": {"session_id": "user_1"}}
)

print(response)
```
